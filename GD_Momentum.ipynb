```python
import pandas as pd

df = pd.read_csv("Position_Salaries.csv.csv")

df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np

X = df["Level"].values
y = df["Salary"].values

plt.scatter(X, y)
plt.xlabel("Level")
plt.ylabel("Salary")
plt.title("Position Salaries Dataset")
plt.show()
```

```python
X_poly = np.column_stack((X, X**2))

X_poly[:5]
```

```python
w1 = 0
w2 = 0
b = 0

lr = 0.0001
epochs = 500

momentum = 0.9

vw1 = 0
vw2 = 0
vb = 0

n = len(X)

losses = []

for epoch in range(epochs):

    y_pred = w1 * X + w2 * (X**2) + b

    loss = np.mean((y - y_pred) ** 2)
    losses.append(loss)

    dw1 = (-2/n) * np.sum(X * (y - y_pred))
    dw2 = (-2/n) * np.sum((X**2) * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    vw1 = momentum * vw1 - lr * dw1
    vw2 = momentum * vw2 - lr * dw2
    vb = momentum * vb - lr * db

    w1 += vw1
    w2 += vw2
    b += vb
```

```python
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GD + Momentum Loss Curve")
plt.show()
```
