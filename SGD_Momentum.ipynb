```python
import pandas as pd

df = pd.read_csv("Position_Salaries.csv.csv")

df.head()
```

```python
import matplotlib.pyplot as plt

X = df["Level"]
y = df["Salary"]

plt.scatter(X, y)
plt.xlabel("Level")
plt.ylabel("Salary")
plt.title("Position Salary Dataset")
plt.show()
```

```python
import numpy as np

X = df["Level"].values
y = df["Salary"].values

X_poly = np.column_stack((X, X**2))

X_poly[:5]
```

```python
w = np.zeros(2)
b = 0

lr = 0.0001
epochs = 500

vw = np.zeros(2)
vb = 0

momentum = 0.9

losses = []

for epoch in range(epochs):

    i = np.random.randint(len(X_poly))

    xi = X_poly[i]
    yi = y[i]

    y_pred = np.dot(xi, w) + b

    error = yi - y_pred

    dw = -2 * xi * error
    db = -2 * error

    vw = momentum * vw - lr * dw
    vb = momentum * vb - lr * db

    w += vw
    b += vb

    total_pred = np.dot(X_poly, w) + b

    loss = np.mean((y - total_pred) ** 2)

    losses.append(loss)
```

```python
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("SGD + Momentum Loss Curve")
plt.show()
```
